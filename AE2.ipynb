```python
import csv

# Empty list to store data
csv_list = []  

while True:
    file_name = input("Enter the CSV file name or enter 'q' to quit the program: ")

    # If input is 'q', exit the program
    if file_name.lower() == 'q':  
        print("Exiting the program.")
        # Exit loop
        break  

    try:
        # Open the CSV file
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.reader(csv_file)
            # Clear csv_list before storing new data in each opening
            csv_list.clear()
            
            # Read and store data using a for loop
            for row in csv_reader:
                # Append each row to the list
                csv_list.append(row)  
            
        print("CSV file loaded successfully.")
        # Exit loop after successfully reading the file
        break  

    # Allow user to try again by showing error
    except FileNotFoundError as e:
        print(f"File does not exist: {e}") 
```

```python
import csv

def retrieve_by_card_category(file_name, card_category):
    results = []  # List to store the matching data

    try:
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.DictReader(csv_file)

            # Iterate over rows and filter based on card category
            for row in csv_reader:
                if row['card_category'] == card_category:
                    results.append({
                        'client_number': row['client_number'],
                        'gender': row['gender'],
                        'education_level': row['education_level'],
                        'income': row['income']
                    })

        return results
    
    except FileNotFoundError:
        print(f"File {file_name} not found.")
        return []

# Example Usage:
file_name = 'customers.csv'
card_category = 'Premium'
data = retrieve_by_card_category(file_name, card_category)
print(data)
```

```python
def retrieve_by_state(file_name, state):
    results = []

    try:
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.DictReader(csv_file)

            for row in csv_reader:
                if row['state'] == state:
                    results.append({
                        'age': row['age'],
                        'gender': row['gender'],
                        'income': row['income'],
                        'card_category': row['card_category']
                    })

        return results

    except FileNotFoundError:
        print(f"File {file_name} not found.")
        return []

# Example Usage:
state = 'California'
data = retrieve_by_state(file_name, state)
print(data)
```

```python
def retrieve_by_marital_status_and_dependents(file_name, marital_status, min_dependents):
    results = []

    try:
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.DictReader(csv_file)

            for row in csv_reader:
                if row['marital_status'] == marital_status and int(row['dependents']) > min_dependents:
                    results.append({
                        'age': row['age'],
                        'gender': row['gender'],
                        'education_level': row['education_level'],
                        'job': row['job']
                    })

        return results

    except FileNotFoundError:
        print(f"File {file_name} not found.")
        return []

# Example Usage:
marital_status = 'Married'
min_dependents = 3
data = retrieve_by_marital_status_and_dependents(file_name, marital_status, min_dependents)
print(data)
```

```python
def retrieve_by_condition(file_name, column1, column2, column3, condition):
    results = []

    try:
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.DictReader(csv_file)

            for row in csv_reader:
                # Apply the condition. Example condition: income > 50000
                if float(row[column1]) > condition:
                    results.append({
                        column1: row[column1],
                        column2: row[column2],
                        column3: row[column3]
                    })

        return results

    except FileNotFoundError:
        print(f"File {file_name} not found.")
        return []

# Example Usage:
column1 = 'income'
column2 = 'gender'
column3 = 'education_level'
condition = 50000  # Condition: Income greater than 50,000
data = retrieve_by_condition(file_name, column1, column2, column3, condition)
print(data)
```
