```python
import csv

# Empty list to store data
csv_list = []  

while True:
    file_name = input("Enter the CSV file name or enter 'q' to quit the program: ")

    # If input is 'q', exit the program
    if file_name.lower() == 'q':  
        print("Exiting the program.")
        # Exit loop
        break  

    try:
        # Open the CSV file
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.reader(csv_file)
            # Clear csv_list before storing new data in each opening
            csv_list.clear()
            
            # Read and store data using a for loop
            for row in csv_reader:
                # Append each row to the list
                csv_list.append(row)  
            
        print("CSV file loaded successfully.")
        # Exit loop after successfully reading the file
        break  

    # Allow user to try again by showing error
    except FileNotFoundError as e:
        print(f"File does not exist: {e}") 



```

```python
import csv

# Create an empty list to store results
card_category = []

# Ask user for the card category
category_input = input("Enter the card category of the customer (Blue/Platinum/Gold/Silver): ")

# Iterate through each row of the CSV list
for row in csv_list[1:]:  # Skip header row
    if row[15] == category_input:  # Check if the row matches the category input
        card_category.append([
            row[0],         # Client_Num
            row[2],         # Gender
            row[4],         # Education_Level
            row[13]         # Income
        ])

# Print the results as a table
if card_category:
    # Print header
    print(f"{'Client_Num':<12} {'Gender':<8} {'Education_Level':<15} {'Income':<10}")
    print("-" * 45)  # Line separator for the table

    # Print rows
    for item in card_category:
        print(f"{item[0]:<12} {item[1]:<8} {item[2]:<15} {item[3]:<10}")
else:
    print(f"Error: No data found for the category '{category_input}'.")
```

```python
#Retrieve the age, gender, income, and credit card category of customers based on their specific state of residence.

residency_state=[]

state = input("Enter customers specific state of residency () : ")

for row in csv_list:
    if row[6] == state:
        residency_state.append({
            'Customer_Age': row[1],
            'Gender': row[2],
            'Income': row[13],
            'Card_Category': row[15]
        })

if residency_state:
    for item in residency_state:
        print (item)

else: 
    print(f'Error')
```

```python
#Retrieve the age, gender, education level, and job of customers who are of a certain marital status and have more than 3 dependents.

marital_and_dependants = []

marital_status = input("What is customers marital status: ")

for row in csv_list:
    if row[5] == marital_status and int(row[3]) >3:
        marital_and_dependants.append({
            'Age': row[1],
            'Gender': row[2],
            'Education_Level': row[4],
            'Customer_Job': row[12]
        })

if marital_and_dependants:
    for item in marital_and_dependants:
        print (item)

else:
    print(f'Error')
```
