```python
import csv

csv_list = []  # Empty list to store data

while True:
    file_name = input("Enter your file name or press enter to quit: ")

    if not file_name:  # If input is empty, exit the program
        print("Exiting the program.")
        break  # Exit loop

    try:
        # Open the CSV file
        with open(file_name, 'r', encoding='UTF-8') as csv_file:
            csv_reader = csv.reader(csv_file)
            
            
            # Read and store data using a for loop
            for row in csv_reader:
                csv_list.append(row)  # Append each row to the list
            
        break  # Exit loop after successfully reading the file

    except FileNotFoundError as e:
        print(f"File does not exist: {e}")  # Allow user to try again
```
