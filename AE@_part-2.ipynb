```python
import pandas as pd

# loads data from a CSV file into memory using the read_csv() function
file_path = "credit_card_customers.csv"

try:
    csv_file = pd.read_csv(file_path, header=0, encoding='utf-8')
    print("CSV file loaded successfully.")
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")

```

```python
# Identify the top 3 occupations of customers with a specific card type and an income exceeding $100,000.

# Get the list of unique card types in the dataset for validation
valid_card_types = csv_file['Card_Category'].unique()

while True:
    # Prompt the user to input a card type
    card_type = input(f"Enter the card type from the following list {list(valid_card_types)} (or type 'quit' to exit): ")
    # Allow user to quit
    if card_type.lower() == 'quit':
        print("Exiting...")
        break

    # Check if the entered card type is valid
    if card_type in valid_card_types:
        # Filter for income > 100000 and selected card type
        filtered_b1 = csv_file.loc[(csv_file['Income'] > 100000) & (csv_file['Card_Category'] == card_type)]

        # Check if any matching records exist
        if not filtered_b1.empty:
            # Get the top 3 occupations by count
            top_3_occupations = filtered_b1['Customer_Job'].value_counts().head(3)
            print("\nTop 3 occupations for the selected card type and income > $100,000:\n")
            print(top_3_occupations)
        else:
            print("\nNo data found for this card type with income over $100,000.")

        break  # Exit the loop after successful processing
    else:
        print("Invalid card type entered. Please try again or type 'quit' to exit.\n")
```

```python
#Analyse the average annual credit card fee and customer income for a specific occupation and card type. 
    
# Get the list of unique card types and occupations for validation
valid_card_types = csv_file['Card_Category'].dropna().unique()
valid_occupations = csv_file['Customer_Job'].dropna().unique()

while True:
    # Prompt the user to input values
    occupation = input(f"Enter the occupation from the list {list(valid_occupations)} (or type 'quit' to exit): ")
    card_type = input(f"Enter the card type from the list {list(valid_card_types)} (or type 'quit' to exit): ")

    # Allow user to quit
    if occupation.lower() == 'quit' or card_type.lower() == 'quit':
        print("Exiting...")
        break

    # Check if both inputs are valid
    if occupation in valid_occupations and card_type in valid_card_types:
        # Filter for the selected occupation and card type
        filtered_b2 = csv_file.loc[
            (csv_file['Customer_Job'] == occupation) & 
            (csv_file['Card_Category'] == card_type)
        ]

        # Check if any matching records exist
        if not filtered_b2.empty:
            # Calculate the average Annual Fee and Income
            avg_fee_income = filtered_b2[['Annual_Fees', 'Income']].mean()
            print(f"\nAverage Annual Fee and Income for occupation '{occupation}' and card type '{card_type}':\n")
            print(avg_fee_income)
        else:
            print("\nNo data found for this combination of occupation and card type.")

        break  # Exit after processing
    else:
        if occupation not in valid_occupations:
            print(f"Invalid occupation: '{occupation}'")
        if card_type not in valid_card_types:
            print(f"Invalid card type: '{card_type}'")
        print("Please try again or type 'quit' to exit.\n")
```
