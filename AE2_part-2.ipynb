```python
import pandas as pd

# loads data from a CSV file into memory using the read_csv() function
file_path = "credit_card_customers.csv"

try:
    csv_file = pd.read_csv(file_path, header=0, encoding='utf-8')
    print("CSV file loaded successfully.")
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")

```

```python
# Get valid card types and jobs
valid_card_types = sorted(csv_file['Card_Category'].dropna().unique())
valid_jobs = sorted(csv_file['Customer_Job'].dropna().unique())

while True:
    # Show card types with numbers
    print("\nAvailable Card Types:")
    for idx, card in enumerate(valid_card_types, 1):
        print(f"{idx}. {card}")
    
    card_input = input("Select a card type by number (or type 'quit' to exit): ").strip()

    # Exit condition
    if card_input.lower() == 'quit':
        print("Exiting B1...")
        break

    # Validate card input
    if not card_input.isdigit() or not (1 <= int(card_input) <= len(valid_card_types)):
        print("Invalid selection. Please try again.")
        continue

    selected_card = valid_card_types[int(card_input) - 1]

    # Filter records with income > 100000 and selected card
    filtered = csv_file[(csv_file['Card_Category'] == selected_card) & (csv_file['Income'] > 100000)]

    if filtered.empty:
        print("No matching records found for this card type with income > 100,000.")
        break

    # Show job titles to select
    print("\nAvailable Job Titles:")
    for idx, job in enumerate(valid_jobs, 1):
        print(f"{idx}. {job}")
    
    job_input = input("Select a job by number to filter further (or type 'quit'): ").strip()

    if job_input.lower() == 'quit':
        print("Exiting B1...")
        break

    if not job_input.isdigit() or not (1 <= int(job_input) <= len(valid_jobs)):
        print("Invalid job selection. Please try again.")
        continue

    selected_job = valid_jobs[int(job_input) - 1]

    # Final filter
    final_filtered = filtered[filtered['Customer_Job'] == selected_job]

    if not final_filtered.empty:
        top_occupations = final_filtered['Customer_Job'].value_counts().head(3)
        print(f"\nTop 3 Occupations for '{selected_card}' card holders with income > $100,000 and job '{selected_job}':")
        print(top_occupations)
    else:
        print("No records found for this job with the selected filters.")

    break
```

```python
#Analyse the average annual credit card fee and customer income for a specific occupation and card type. 
    
# Get valid occupations and card types
valid_occupations = csv_file['Customer_Job'].dropna().unique()
valid_card_types = csv_file['Card_Category'].dropna().unique()

while True:
    # Input for occupation and card type
    print(f"\nAvailable Occupations: {list(valid_occupations)}")
    occupation = input("Enter an occupation (or type 'quit' to exit): ").strip()
    
    print(f"\nAvailable Card Types: {list(valid_card_types)}")
    card_type = input("Enter a card type (or type 'quit' to exit): ").strip()

    # Exit condition
    if occupation.lower() == 'quit' or card_type.lower() == 'quit':
        print("Exiting B2...")
        break

    # Validate inputs
    if occupation in valid_occupations and card_type in valid_card_types:
        # Filter and calculate averages
        filtered = csv_file[(csv_file['Customer_Job'] == occupation) & 
                            (csv_file['Card_Category'] == card_type)]

        if not filtered.empty:
            avg_values = filtered[['Annual_Fees', 'Income']].mean()
            print(f"\nAverage values for Occupation: '{occupation}' and Card Type: '{card_type}':")
            print(avg_values)
        else:
            print("No matching records found.")
        break
    else:
        print("Invalid input. Please ensure both occupation and card type are from the list.")
```

```python
# Identify the average annual fee for each marital status category among customers with a specific education level, homeownership status, and income above a certain threshold.

# Get unique education levels and home ownership types
education_levels = csv_file['Education_Level'].dropna().unique()
home_ownership = csv_file['House_Owner'].dropna().unique()

while True:
    # Education level selection
    print("\nAvailable Education Levels:")
    for idx, level in enumerate(education_levels, 1):
        print(f"{idx}. {level}")
    edu_input = input("Select Education Level by number (or 'quit'): ").strip()

    if edu_input.lower() == 'quit':
        print("Exiting B3...")
        break
    if not edu_input.isdigit() or not (1 <= int(edu_input) <= len(education_levels)):
        print("Invalid input. Try again.")
        continue

    selected_edu = education_levels[int(edu_input) - 1]

    # Home ownership and income threshold
    home_input = input("Do you own a home? (yes/no or 'quit'): ").strip().lower()
    if home_input == 'quit':
        print("Exiting B3...")
        break
    if home_input not in ['yes', 'no']:
        print("Invalid home ownership input. Try again.")
        continue

    income_input = input("Enter minimum income threshold (numeric or 'quit'): ").strip()
    if income_input.lower() == 'quit':
        print("Exiting B3...")
        break
    try:
        income_threshold = float(income_input)
    except ValueError:
        print("Invalid income input. Please enter a number.")
        continue

    # Apply filters
    filtered = csv_file[
        (csv_file['Education_Level'] == selected_edu) &
        (csv_file['House_Owner'].str.lower() == home_input) &
        (csv_file['Income'] > income_threshold)
    ]

    if filtered.empty:
        print("No records found for the given filters.")
        continue

    # Filter by marital status
    print(f"\nFiltered records found: {len(filtered)}")
    print(f"Available Marital Statuses: {filtered['Marital_Status'].unique()}")

    marital_input = input("Enter a marital status from the above (or 'quit'): ").strip()
    if marital_input.lower() == 'quit':
        print("Exiting B3...")
        break

    final_filtered = filtered[filtered['Marital_Status'] == marital_input]

    if final_filtered.empty:
        print("No records for this marital status.")
    else:
        # Calculate average annual fee
        avg_fee = final_filtered['Annual_Fees'].mean()
        print(f"\nAverage Annual Fee for '{marital_input}' customers:")
        print(f"${avg_fee:.2f}")
        print("\nDetailed View:")
        print(final_filtered[['Marital_Status', 'Annual_Fees', 'Education_Level', 'House_Owner', 'Income']])

    # Retry loop
    retry = input("\nDo you want to try another input? (yes/no): ").strip().lower()
    if retry != 'yes':
        print("Exiting B3...")
        break
```
