```python
import pandas as pd

# loads data from a CSV file into memory using the read_csv() function
file_path = "credit_card_customers.csv"

try:
    df_ccards = pd.read_csv(file_path, header=0, encoding='utf-8')
    print("Credit Card Customers data frame loaded successfully.")
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")
```

```python
# Get valid card types and jobs
valid_card_types = sorted(df_ccards['Card_Category'].dropna().unique())
valid_jobs = sorted(df_ccards['Customer_Job'].dropna().unique())

while True:
    # Show card types with numbers
    print("\nAvailable Card Types:")
    for idx, card in enumerate(valid_card_types, 1):
        print(f"{idx}. {card}")
    
    card_input = input("Select a card type by number (or type 'quit' to exit): ").strip()

    # Exit condition
    if card_input.lower() == 'quit':
        print("Exiting B1...")
        break

    # Validate card input
    if not card_input.isdigit() or not (1 <= int(card_input) <= len(valid_card_types)):
        print("Invalid selection. Please try again.")
        continue

    selected_card = valid_card_types[int(card_input) - 1]

    # Filter records with income > 100000 and selected card
    filtered = df_ccards.loc[(df_ccards['Card_Category'] == selected_card) & (df_ccards['Income'] > 100000)]

    if filtered.empty:
        print("No matching records found for this card type with income > 100,000.")
        break

    else:
        # Get top 3 job categories
        top_jobs = filtered['Customer_Job'].value_counts().head(3)
        print(f"\nTop 3 Occupations for '{selected_card}' card holders with income > $100,000:")
        print(top_jobs)

    # Retry loop
    again = input("\nWould you like to run this analysis again? (yes/no): ").strip().lower()
    if again != 'yes':
        print("Exiting B1...")
        break
```

```python
#Analyse the average annual credit card fee and customer income for a specific occupation and card type. 
    
# Get valid occupations and card types
valid_occupations = df_ccards['Customer_Job'].dropna().unique()
valid_card_types = df_ccards['Card_Category'].dropna().unique()

while True:
    # Input for occupation and card type
    print("\nAvailable Occupations:")
    for idx, occupation in enumerate(valid_occupations, 1):
        print(f"{idx}. {occupation}")
    
    occupation = input("Enter an occupation (or type 'quit' to exit): ").strip()

    # Exit condition
    if occupation.lower() == 'quit':
        print("Exiting B1...")
        break

    #validate occupation input 
    if not occupation.isdigit() or not (1 <= int(occupation) <= len(valid_occupations)):
        print("Invalid selection. Please try again.")
        continue

    selected_occupation = valid_occupations[int(occupation) - 1]

    #Input for card types 
    print("\nAvailable Card Types:")
    for idx, card in enumerate(valid_card_types, 1):
        print(f"{idx}. {card}")
    
    card_input = input("Enter a card type (or type 'quit' to exit): ").strip()

    # Exit condition
    if card_input.lower() == 'quit':
        print("Exiting B2...")
        break

    # Validate card input
    if not card_input.isdigit() or not (1 <= int(card_input) <= len(valid_card_types)):
        print("Invalid selection. Please try again.")
        continue

    selected_card = valid_card_types[int(card_input) - 1]

    # Filter and calculate averages
    filtered = df_ccards.loc[(df_ccards['Customer_Job'] == selected_occupation) & (df_ccards['Card_Category'] == selected_card)]
    if not filtered.empty:
        avg_values = filtered[['Annual_Fees', 'Income']].mean()
        print(f"\nAverage annual credit card fee for Occupation: '{selected_occupation}' and Card Type: '{selected_card}':")
        print(avg_values)
    else:
        print("No matching records found.")

    # Retry loop
    again = input("\nWould you like to run this analysis again? (yes/no): ").strip().lower()
    if again != 'yes':
        print("Exiting B2...")
        break
    
```

```python
# Identify the average annual fee for each marital status category among customers with a specific education level, homeownership status, and income above a certain threshold.

# Get unique education levels and home ownership types
education_levels = df_ccards['Education_Level'].dropna().unique()
home_ownership = df_ccards['House_Owner'].dropna().unique()
marital_status = df_ccards['Marital_Status'].dropna().unique()

while True:
    # Education level selection
    print("\nAvailable Education Levels:")
    for idx, level in enumerate(education_levels, 1):
        print(f"{idx}. {level}")
    
    edu_input = input("Select Education Level by number (or 'quit'): ").strip()

    if edu_input.lower() == 'quit':
        print("Exiting B3...")
        break
    
    #Vaalidate edicational input
    if not edu_input.isdigit() or not (1 <= int(edu_input) <= len(education_levels)):
        print("Invalid input. Try again.")
        continue

    selected_edu = education_levels[int(edu_input) - 1]

    # Home ownership and income threshold
    home_input = input("Do you own a home? (yes/no or 'quit'): ").strip().lower()
    if home_input == 'quit':
        print("Exiting B3...")
        break
    if home_input not in ['yes', 'no']:
        print("Invalid home ownership input. Try again.")
        continue

    income_input = input("Enter minimum income threshold (numeric or 'quit'): ").strip()
    if income_input.lower() == 'quit':
        print("Exiting B3...")
        break
    try:
        income_threshold = float(income_input)
    except ValueError:
        print("Invalid income input. Please enter a number.")
        continue

    # Apply filters
    filtered = df_ccards.loc[(df_ccards['Education_Level'] == selected_edu) & (df_ccards['House_Owner'].str.lower() == home_input) & (df_ccards['Income'] > income_threshold)]

    if filtered.empty:
        print("No records found for the given filters.")
        continue

    # Filter by marital status
    print(f"\nFiltered records found: {len(filtered)}")
    
    print(f"\nAvailable Marital Statuses:")
    for idx, status in enumerate (marital_status ,1):
        print(f"{idx}. {status}")

    marital_input = input("Enter a marital status from the above (or 'quit'): ").strip()
    if marital_input.lower() == 'quit':
        print("Exiting B3...")
        break
    
    if not marital_input.isdigit() or not (1 <= int(marital_input) <= len(marital_status)):
        print("Invalid input. Try again.")
        continue
    
    selected_status = marital_status[int(marital_input) -1]

    final_filtered = filtered[filtered['Marital_Status'] == selected_status]

    if final_filtered.empty:
        print("No records for this marital status.")
    else:
        # Calculate average annual fee
        avg_fee = final_filtered['Annual_Fees'].mean()
        print(f"\nAverage Annual Fee for '{selected_status}' customers:")
        print(f"${avg_fee:.2f}")

    # Retry loop
    retry = input("\nDo you want to try another input? (yes/no): ").strip().lower()
    if retry != 'yes':
        print("Exiting B3...")
        break
```

```python
# Prepare lists for inputs
employment_statuses = df_ccards['Customer_Job'].dropna().unique()
expense_types = df_ccards['Exp Type'].dropna().unique()

while True:
    # Select employment status
    print("\nEmployment Status Options:")
    for i, job in enumerate(employment_statuses, 1):
        print(f"{i}. {job}")
        
    emp_input = input("Select Employment Status (number or 'quit'): ").strip()
    if emp_input.lower() == 'quit':
        break
    if not emp_input.isdigit() or not (1 <= int(emp_input) <= len(employment_statuses)):
        print("Invalid input. Try again.")
        continue
        
    selected_job = employment_statuses[int(emp_input) - 1]

    # Loan and car ownership inputs
    loan_input = input("Do you have a loan? (Yes/No): ").strip().lower()
    car_input = input("Do you own a car? (Yes/No): ").strip().lower()
    if loan_input not in ['yes', 'no'] or car_input not in ['yes', 'no']:
        print("Invalid loan or car input. Use Yes/No.")
        continue

    # Expense type selection
    print("\nExpense Type Options:")
    for i, etype in enumerate(expense_types, 1):
        print(f"{i}. {etype}")
    exp_input = input("Select Expense Type (number or 'quit'): ").strip()
   
    if exp_input.lower() == 'quit':
        break
    if not exp_input.isdigit() or not (1 <= int(exp_input) <= len(expense_types)):
        print("Invalid expense type.")
        continue
   
    selected_exp = expense_types[int(exp_input) - 1]

    # Apply all filters
    custom_filtered = df_ccards.loc[(df_ccards['Customer_Job'] == selected_job) & (df_ccards['Personal_loan'] == loan_input) & (df_ccards['Car_Owner'] == car_input) & (df_ccards['Exp Type'] == selected_exp)]

    # Display insight
    if custom_filtered.empty:
        print("No matching records found.")
    else:
        avg_interest = custom_filtered['Interest_Earned'].mean()
        avg_limit = custom_filtered['Credit_Limit'].mean()
        print(f"\n Average Interest Earned: ${avg_interest:.2f}")
        print(f" Average Credit Limit: ${avg_limit:.2f}")

    # Retry loop
    again = input("\nWould you like to run this analysis again? (yes/no): ").strip().lower()
    if again != 'yes':
        print("Exiting B4...")
        break
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load the data
def load_data(file_path):
    df = pd.read_csv(file_path, encoding='utf-8')
    print("Data Frame is loaded")
    return df

# ------------------------------
# Task c1: Proportion of each card type for a specific education level
# ------------------------------

def select_education_level(df):
    education_levels = df['Education_Level'].dropna().unique()
    while True:
        print("\nSelect an Education Level (or type 0 to return to the main menu):")
        for idx, level in enumerate(education_levels):
            print(f"{idx + 1}. {level}")
        print("0. Return to Main Menu")

        try:
            choice = int(input("Enter the number corresponding to your choice: "))
            if choice == 0:
                return None
            elif 1 <= choice <= len(education_levels):
                return education_levels[choice - 1]
            else:
                print("Invalid number. Please select a valid option.")
        except ValueError:
            print("Invalid input. Please enter a number.")

def plot_card_type_pie_chart(df, selected_level):
    filtered_df = df[df['Education_Level'] == selected_level]

    if filtered_df.empty:
        print(f"No data available for {selected_level}.")
        return

    card_counts = filtered_df['Card_Category'].value_counts()

    plt.figure()
    plt.pie(card_counts.values, labels=card_counts.index, autopct='%1.1f%%', startangle=90)
    plt.title(f'Proportion of Card Types for {selected_level}')
    plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1))
    plt.show()

def task_c1(df):
    selected_level = select_education_level(df)
    if selected_level is None:
        return
    plot_card_type_pie_chart(df, selected_level)

# ------------------------------
# Task c2: Monthly trend of average interest earned for each card category
# ------------------------------

def plot_monthly_interest_trends(df):
    if 'Week_Num' not in df.columns:
        print("Week_Num column is missing!")
        return

    df['Week_Num'] = df['Week_Num'].str.replace('Week-', '').astype(int)
    df['Month'] = df['Week_Num'] // 4 + 1

    monthly_avg_interest = df.groupby(['Month', 'Card_Category'])['Interest_Earned'].mean().unstack()

    plt.figure(figsize=(10, 6))
    for card_cat in monthly_avg_interest.columns:
        plt.plot(monthly_avg_interest.index, monthly_avg_interest[card_cat], marker='o', label=card_cat)

    plt.title('Monthly Trend of Average Interest Earned by Card Category')
    plt.xlabel('Month')
    plt.ylabel('Average Interest Earned')
    plt.legend(title='Card Category')
    plt.grid(True)
    plt.xticks(range(1, 13))
    plt.tight_layout()
    plt.show()

def task_c2(df):
    plot_monthly_interest_trends(df)

# ------------------------------
# Task c3: Compare average annual fees and customer acquisition costs
# ------------------------------

def plot_fees_vs_acquisition(df):
    avg_fees = df.groupby('Card_Category')['Annual_Fees'].mean()
    avg_acquisition = df.groupby('Card_Category')['Customer_Acq_Cost'].mean()

    card_categories = avg_fees.index
    x_axis = range(len(card_categories))
    bar_width = 0.4

    plt.figure(figsize=(10, 6))
    plt.bar([x - 0.2 for x in x_axis], avg_fees.values, width=bar_width, label='Annual Fees', color='purple')
    plt.bar([x + 0.2 for x in x_axis], avg_acquisition.values, width=bar_width, label='Acquisition Cost', color='orange')

    plt.title('Average Annual Fees vs Acquisition Costs by Card Category')
    plt.xlabel('Card Category')
    plt.ylabel('Amount ($)')
    plt.xticks(x_axis, card_categories)
    plt.legend()
    plt.tight_layout()
    plt.show()

def task_c3(df):
    plot_fees_vs_acquisition(df)

# ------------------------------
# Task c4: Expenses Type per Card Category (Stacked Bar Chart)
# ------------------------------

def plot_expense_type_for_card_categories_stacked(df):
    grouped = df.groupby(['Exp Type', 'Card_Category']).size().unstack(fill_value=0)

    exp_types = grouped.index
    card_categories = grouped.columns

    x_axis = np.arange(len(exp_types))

    plt.figure(figsize=(10, 6))

    bottom = np.zeros(len(exp_types))

    for card_cat in card_categories:
        plt.bar(x_axis, grouped[card_cat], width=0.5, label=card_cat, bottom=bottom)
        bottom += grouped[card_cat].values

    plt.xticks(x_axis, exp_types, rotation=45)
    plt.xlabel("Expense Type")
    plt.ylabel("Number of Customers")
    plt.title("Number of Customers per Expense Type (Stacked by Card Category)")
    plt.legend(title="Card Category")
    plt.tight_layout()
    plt.show()

def task_c4(df):
    plot_expense_type_for_card_categories_stacked(df)

# ------------------------------
# Main Menu
# ------------------------------

def main_menu(df):
    while True:
        print("\n=== Main Menu ===")
        print("1. Proportion of each card type for a specific education level (Task c1)")
        print("2. Monthly trend of average interest earned for each card category (Task c2)")
        print("3. Compare average annual fees and acquisition costs across card categories (Task c3)")
        print("4. Expenses Type per Card Category (Stacked Bar Chart) (Task c4)")
        print("0. Quit")

        try:
            choice = int(input("Select an option (0-4): "))
            if choice == 0:
                print("\nThank you for using our customer data explorer. Goodbye!!")
                break
            elif choice == 1:
                task_c1(df)
            elif choice == 2:
                task_c2(df)
            elif choice == 3:
                task_c3(df)
            elif choice == 4:
                task_c4(df)
            else:
                print("Invalid option. Please select a number between 0 and 4.")
        except ValueError:
            print("Invalid input. Please enter a number.")

# ------------------------------
# Main Function
# ------------------------------

def main():
    file_path = "credit_card_customers.csv"
    df_ccard = load_data(file_path)
    main_menu(df_ccard)

if __name__ == "__main__":
    main()
```
